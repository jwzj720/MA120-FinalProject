## High level goals: visualize the word-2-vec space by walking through PCA. 
### Visualize the word-2-vec space in 2D and 3D, then look at semantic meanings in the space and talk about how the model assigns meaning to words

```python
wv = api.load('word2vec-google-news-300')
```

```python
for index, word in enumerate(wv.index_to_key):
    if index == 10:
        break
    print(f"word #{index}/{len(wv.index_to_key)} is {word}")
```

```python
import numpy as np
```

```python
plot_words = ["king", "queen", "man", "woman", "glass","boy","girl"]
```

```python
word_matrix = np.array([wv[word] for word in plot_words])
```

```python
num_components = 3
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
pca = PCA(n_components=num_components)
```

```python
pca_result = pca.fit_transform(word_matrix)
pca_result
```

```python
data = {
    'x': pca_result[:, 0],
    'y': pca_result[:, 1],
    'z': pca_result[:, 2],
    'word': plot_words
}

# Create an interactive 3D scatter plot
fig = px.scatter_3d(data, x='x', y='y', z='z', text='word')

# Customize the appearance of the plot (optional)
fig.update_traces(marker=dict(size=8))

# Show the interactive plot
fig.show()
```
